````python
### Solution ###

def selection_sort(scores):
    n = len(scores)
    for i in range(n - 1):
        # Sol: 미정렬 영역의 첫 위치를 최댓값 후보로 정합니다.
        max_index = i
        for j in range(i + 1, n):
            if scores[j] > scores[max_index]:
                max_index = j
        # Sol: 선택한 최댓값을 현재 자리로 옮깁니다.
        if max_index != i:
            scores[i], scores[max_index] = scores[max_index], scores[i]


scores = list(map(int, input("부스 점수(공백 구분): ").split()))
selection_sort(scores)
print(f"정렬된 점수: {scores}")
````

````python
### Solution ###

def insertion_sort(records):
    for i in range(1, len(records)):
        key = records[i]
        j = i - 1
        # Sol: 기록이 더 큰 항목만 이동해 동점자의 순서를 유지합니다.
        while j >= 0 and records[j][1] > key[1]:
            records[j + 1] = records[j]
            j -= 1
        # Sol: 이름과 기록을 함께 빈 자리에 넣습니다.
        records[j + 1] = key


count = int(input("참가자 수: "))
records = []
for i in range(count):
    name, seconds = input("이름과 기록: ").split()
    records.append((name, int(seconds)))
insertion_sort(records)
for name, seconds in records:
    print(name, seconds)
````

### 도전문제 4. 정렬 방법의 비교 횟수 조사


<img src="../images/week6/week6_chal4.png" width="600" alt="두 정렬 방법을 비교하며 공부하는 학생들">


**문제 상황**

같은 데이터라도 정렬 방법에 따라 필요한 연산 수가 달라집니다.
교환이 없는 차례에서 멈추는 **조기 종료 버블 정렬**과 **선택 정렬**의 비교 횟수를 조사하세요.
두 방법 모두 오름차순으로 정렬합니다. 실행 시간 대신 정확히 셀 수 있는 원소 비교 횟수를 출력합니다.

**만들 함수와 제공 코드**

| 함수 | 역할 |
| --- | --- |
| `bubble_count(values)` | 리스트 자체를 조기 종료 버블 정렬하고, 원소 비교 횟수를 정수로 반환합니다. 학생이 작성합니다. |
| `selection_count(values)` | 리스트 자체를 선택 정렬하고, 원소 비교 횟수를 반환합니다. 완성 코드가 제공됩니다. |

**비교 횟수의 기준**

- `values[j] > values[j + 1]` 또는 `values[j] < values[min_index]`처럼 **원소 값 두 개를 비교할 때마다 1회**로 셉니다. 조건이 거짓이어도 1회입니다.
- 반복 범위, 인덱스, `swapped`의 검사와 교환, 대입은 비교 횟수에 포함하지 않습니다.
- 버블 정렬은 각 차례마다 왼쪽부터 인접한 값을 비교하고, 끝난 뒤 오른쪽 비교 범위를 하나 줄입니다.
- 한 차례 동안 교환이 한 번도 없었으면 그 차례의 비교까지 센 뒤 종료합니다.
- 비교할 원소가 2개 미만이면 두 방법 모두 0회입니다.

**입출력 형식**

- **입력:** 정수를 공백으로 구분한 한 줄입니다. 아무것도 입력하지 않고 Enter를 누르면 빈 리스트입니다.
- **출력:** `버블 결과`, `버블 비교`, `선택 결과`, `선택 비교`를 예시처럼 네 줄로 출력합니다.

**조건과 작성 안내**

- 정수는 0개 이상 8개 이하이며 각 값은 0 이상 100 이하입니다. 중복도 가능합니다.
- 두 함수에 **같은 원본의 서로 다른 복사본**을 전달합니다. 첫 번째로 정렬한 결과를 두 번째 함수의 입력으로 사용하지 않습니다.
- 복사와 출력 코드는 제공됩니다. `bubble_count()`의 본문만 완성하세요. `sort()`와 `sorted()`는 사용하지 않습니다.
- 비교 횟수는 이 문제에서 정한 한 종류의 기본 연산입니다. 이것만으로 모든 입력에서 실제 실행 시간이 더 빠르다고 결론 내리지는 않습니다.

**입출력 예시**

입력
```
1 2 3 4
```

출력
```

버블 결과: [1, 2, 3, 4]
버블 비교: 3회
선택 결과: [1, 2, 3, 4]
```

````python
### Solution ###

def bubble_count(values):
    comparisons = 0
    for end in range(len(values) - 1, 0, -1):
        swapped = False
        for j in range(end):
            # Sol: 교환 여부와 관계없이 원소 비교 한 번을 셉니다.
            comparisons += 1
            if values[j] > values[j + 1]:
                values[j], values[j + 1] = values[j + 1], values[j]
                swapped = True
        # Sol: 교환이 없는 차례까지 센 뒤 반복을 종료합니다.
        if not swapped:
            break
    return comparisons


def selection_count(values):
    comparisons = 0
    for i in range(len(values) - 1):
        min_index = i
        for j in range(i + 1, len(values)):
            comparisons += 1
            if values[j] < values[min_index]:
                min_index = j
        if min_index != i:
            values[i], values[min_index] = values[min_index], values[i]
    return comparisons


values = list(map(int, input("정수 목록(공백 구분): ").split()))
bubble_values = values.copy()
selection_values = values.copy()
bubble_comparisons = bubble_count(bubble_values)
selection_comparisons = selection_count(selection_values)
print(f"버블 결과: {bubble_values}")
print(f"버블 비교: {bubble_comparisons}회")
print(f"선택 결과: {selection_values}")
print(f"선택 비교: {selection_comparisons}회")
````

````python
### Solution ###

def rank_entries(entries):
    for i in range(1, len(entries)):
        key = entries[i]
        j = i - 1
        # Sol: 점수가 더 낮은 항목만 이동해 동점 순서를 유지합니다.
        while j >= 0 and entries[j][1] < key[1]:
            entries[j + 1] = entries[j]
            j -= 1
        # Sol: 현재 참가자를 점수 내림차순의 알맞은 자리에 넣습니다.
        entries[j + 1] = key


count = int(input("참가자 수: "))
entries = []
for i in range(count):
    name, score = input("이름과 점수: ").split()
    entries.append((name, int(score)))
k = int(input("선정 인원: "))
rank_entries(entries)
if k == 0:
    print("선발된 참가자가 없습니다.")
else:
    for i in range(k):
        print(entries[i][0], entries[i][1])
````
