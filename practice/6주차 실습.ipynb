````python
### Solution ###

# 전달받은 무게 리스트 자체를 정렬하는 함수입니다.
def bubble_sort(weights):
    # Sol: 상자 수를 저장해 비교 범위를 정합니다.
    n = len(weights)
    # Sol: 한 차례가 끝날 때마다 오른쪽 비교 범위를 줄입니다.
    for end in range(n - 1, 0, -1):
        # Sol: 왼쪽부터 인접한 두 무게를 차례로 확인합니다.
        for j in range(end):
            # Sol: 왼쪽 무게가 더 클 때만 교환이 필요합니다.
            if weights[j] > weights[j + 1]:
                # Sol: 큰 무게를 오른쪽으로 보내도록 두 값을 교환합니다.
                weights[j], weights[j + 1] = weights[j + 1], weights[j]


# 입력한 무게를 공백으로 나누고 정수 리스트로 변환합니다.
weights = list(map(int, input("상자 무게(공백 구분): ").split()))
# 구현한 버블 정렬 함수를 호출해 무게순으로 정리합니다.
bubble_sort(weights)
# 정렬이 끝난 무게 리스트를 출력합니다.
print(f"정렬된 무게: {weights}")
````

````python
### Solution ###

# 전달받은 점수 리스트 자체를 정렬하는 함수입니다.
def selection_sort(scores):
    # Sol: 점수의 개수를 저장해 탐색 범위를 정합니다.
    n = len(scores)
    # Sol: 가장 앞자리부터 높은 점수를 하나씩 확정합니다.
    for i in range(n - 1):
        # Sol: 미정렬 영역의 첫 위치를 최댓값 후보로 정합니다.
        max_index = i
        # Sol: 후보 다음 위치부터 남은 점수를 모두 확인합니다.
        for j in range(i + 1, n):
            # Sol: 현재 후보보다 높은 점수가 있는지 비교합니다.
            if scores[j] > scores[max_index]:
                # Sol: 더 높은 점수를 찾으면 후보 위치를 갱신합니다.
                max_index = j
        # Sol: 최댓값이 현재 자리와 다른 위치에 있을 때만 교환합니다.
        if max_index != i:
            # Sol: 선택한 최댓값을 현재 자리로 옮깁니다.
            scores[i], scores[max_index] = scores[max_index], scores[i]


# 입력한 점수를 공백으로 나누고 정수 리스트로 변환합니다.
scores = list(map(int, input("부스 점수(공백 구분): ").split()))
# 구현한 선택 정렬 함수를 호출해 높은 점수부터 정리합니다.
selection_sort(scores)
# 정렬이 끝난 점수 리스트를 출력합니다.
print(f"정렬된 점수: {scores}")
````

````python
### Solution ###

# 리스트를 정렬하고 원소 비교 횟수를 반환하는 함수입니다.
def bubble_count(values):
    # Sol: 이번 정렬에서 수행할 원소 비교 횟수를 0으로 시작합니다.
    comparisons = 0
    # Sol: 한 차례가 끝날 때마다 오른쪽 비교 범위를 줄입니다.
    for end in range(len(values) - 1, 0, -1):
        # Sol: 이번 차례의 교환 여부를 새로 기록합니다.
        swapped = False
        # Sol: 왼쪽부터 인접한 두 원소를 차례로 확인합니다.
        for j in range(end):
            # Sol: 바로 다음 원소 비교를 참과 거짓에 관계없이 한 번 셉니다.
            comparisons += 1
            # Sol: 왼쪽 값이 더 커서 순서를 바꿔야 하는지 비교합니다.
            if values[j] > values[j + 1]:
                # Sol: 인접한 두 값을 교환해 큰 값을 오른쪽으로 보냅니다.
                values[j], values[j + 1] = values[j + 1], values[j]
                # Sol: 이번 차례에 교환이 있었음을 표시합니다.
                swapped = True
        # Sol: 이번 차례 전체에서 교환이 없었는지 확인합니다.
        if not swapped:
            # Sol: 이미 정렬되었으므로 남은 차례를 생략합니다.
            break
    # Sol: 교환 없는 마지막 차례까지 포함한 원소 비교 횟수를 반환합니다.
    return comparisons


# 비교 대상으로 사용할 선택 정렬 함수는 완성된 코드로 제공됩니다.
def selection_count(values):
    # 선택 정렬에서 수행한 원소 비교 횟수를 셀 변수를 준비합니다.
    comparisons = 0
    # 가장 앞자리부터 작은 값을 하나씩 확정합니다.
    for i in range(len(values) - 1):
        # 미정렬 영역의 첫 위치를 최솟값 후보로 정합니다.
        min_index = i
        # 후보 다음 위치부터 남은 값을 모두 확인합니다.
        for j in range(i + 1, len(values)):
            # 바로 다음 원소 비교를 참과 거짓에 관계없이 한 번 셉니다.
            comparisons += 1
            # 현재 값이 최솟값 후보보다 작은지 비교합니다.
            if values[j] < values[min_index]:
                # 더 작은 값을 찾으면 후보 위치를 갱신합니다.
                min_index = j
        # 선택한 최솟값이 현재 자리와 다른 위치에 있는지 확인합니다.
        if min_index != i:
            # 최솟값을 현재 자리로 옮기도록 두 값을 교환합니다.
            values[i], values[min_index] = values[min_index], values[i]
    # 선택 정렬에서 수행한 원소 비교 횟수를 반환합니다.
    return comparisons


# 입력을 정수 리스트로 변환합니다. 빈 줄은 빈 리스트가 됩니다.
values = list(map(int, input("정수 목록(공백 구분): ").split()))
# 버블 정렬에 전달할 독립적인 복사본을 만듭니다.
bubble_values = values.copy()
# 선택 정렬에도 같은 원본의 별도 복사본을 준비합니다.
selection_values = values.copy()

# 버블 정렬을 실행하고 반환된 비교 횟수를 저장합니다.
bubble_comparisons = bubble_count(bubble_values)
# 선택 정렬을 실행하고 반환된 비교 횟수를 저장합니다.
selection_comparisons = selection_count(selection_values)

# 버블 정렬이 수정한 리스트를 출력합니다.
print(f"버블 결과: {bubble_values}")
# 버블 정렬의 원소 비교 횟수를 출력합니다.
print(f"버블 비교: {bubble_comparisons}회")
# 선택 정렬이 수정한 리스트를 출력합니다.
print(f"선택 결과: {selection_values}")
# 선택 정렬의 원소 비교 횟수를 출력합니다.
print(f"선택 비교: {selection_comparisons}회")
````

````python
### Solution ###

# 접수된 참가자를 점수순으로 정렬하는 함수입니다.
def rank_entries(entries):
    # Sol: 첫 항목은 정렬된 것으로 보고 두 번째 항목부터 삽입합니다.
    for i in range(1, len(entries)):
        # Sol: 삽입할 참가자의 이름과 점수를 함께 보관합니다.
        key = entries[i]
        # Sol: 정렬된 앞부분의 마지막 위치부터 확인합니다.
        j = i - 1
        # Sol: 앞쪽 점수가 더 낮을 때만 이동해 동점자의 순서를 유지합니다.
        while j >= 0 and entries[j][1] < key[1]:
            # Sol: 점수가 낮은 참가자를 오른쪽으로 한 칸 옮깁니다.
            entries[j + 1] = entries[j]
            # Sol: 그보다 한 칸 왼쪽의 참가자를 이어서 확인합니다.
            j -= 1
        # Sol: 보관한 참가자를 점수 내림차순의 알맞은 자리에 넣습니다.
        entries[j + 1] = key


# 입력할 참가자 수를 정수로 읽습니다.
count = int(input("참가자 수: "))
# 참가자의 이름과 점수를 저장할 리스트를 준비합니다.
entries = []

# 참가자 수만큼 이름과 점수를 접수 순서대로 읽습니다.
for i in range(count):
    # 한 줄의 입력을 이름과 점수로 나눕니다.
    name, score = input("이름과 점수: ").split()
    # 점수를 정수로 변환하고 이름과 함께 튜플로 저장합니다.
    entries.append((name, int(score)))

# 선정할 인원을 정수로 읽습니다.
k = int(input("선정 인원: "))
# 구현한 삽입 정렬 함수로 높은 점수부터 정리합니다.
rank_entries(entries)

# 선정할 인원이 0명인지 확인합니다.
if k == 0:
    # 선정 인원이 없을 때 안내 문구를 출력합니다.
    print("선발된 참가자가 없습니다.")
# 선정할 인원이 있으면 정렬된 앞부분을 출력합니다.
else:
    # 앞에서부터 정확히 k명의 참가자를 선택합니다.
    for i in range(k):
        # 선정된 참가자의 이름과 점수를 한 줄로 출력합니다.
        print(entries[i][0], entries[i][1])
````
