````python
### Solution ###

class Stack:
    def __init__(self):
        # Sol: 이 스택의 작업을 저장할 빈 리스트를 만듭니다.
        self.items = []

    def push(self, item):
        # Sol: 새 작업을 스택의 맨 위에 추가합니다.
        self.items.append(item)

    def pop(self):
        # Sol: 작업이 있으면 맨 위 작업을 제거하고 반환합니다.
        if self.isEmpty():
            return None
        return self.items.pop()

    def peek(self):
        # Sol: 작업을 제거하지 않고 맨 위 작업을 반환합니다.
        if self.isEmpty():
            return None
        return self.items[-1]

    def isEmpty(self):
        # Sol: 저장된 작업이 없는지 반환합니다.
        return len(self.items) == 0


stack = Stack()
count = int(input("명령 수: "))
for i in range(count):
    parts = input("명령: ").split()
    command = parts[0]
    if command == "추가":
        # Sol: 입력한 작업을 스택에 추가합니다.
        stack.push(parts[1])
    elif command == "취소":
        # Sol: 최근 작업을 제거해 item에 저장합니다.
        item = stack.pop()
        if item is None:
            print("작업이 없습니다.")
        else:
            print(f"취소: {item}")
    elif command == "확인":
        # Sol: 최근 작업을 제거하지 않고 item에 저장합니다.
        item = stack.peek()
        if item is None:
            print("작업이 없습니다.")
        else:
            print(f"최근 작업: {item}")
````

````python
### Solution ###
from collections import deque

# Sol: 양쪽 끝을 사용할 빈 보관대를 만듭니다.
shelf = deque()

count = int(input("명령 수: "))
for i in range(count):
    parts = input("명령: ").split()
    command = parts[0]
    if command == "왼쪽추가":
        # Sol: 새 물품을 왼쪽 끝에 추가합니다.
        shelf.appendleft(parts[1])
    elif command == "오른쪽추가":
        # Sol: 새 물품을 오른쪽 끝에 추가합니다.
        shelf.append(parts[1])
    elif command == "왼쪽꺼내기":
        # Sol: 왼쪽 끝 물품을 꺼내고 빈 상태는 따로 안내합니다.
        if shelf:
            item = shelf.popleft()
            print(f"꺼낸 물품: {item}")
        else:
            print("보관대가 비어 있습니다.")
    elif command == "오른쪽꺼내기":
        # Sol: 오른쪽 끝 물품을 꺼내고 빈 상태는 따로 안내합니다.
        if shelf:
            item = shelf.pop()
            print(f"꺼낸 물품: {item}")
        else:
            print("보관대가 비어 있습니다.")
````

````python
### Solution ###

# 방문지 하나와 다음 방문지의 참조를 저장합니다.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# 입력 목록으로 초기 경로를 만듭니다.
def make_route(names):
    head = None
    tail = None
    for name in names:
        node = Node(name)
        if head is None:
            head = node
        else:
            tail.next = node
        tail = node
    return head


def insert_after(head, target, name):
    # Sol: 다음 노드를 따라 기준 방문지를 찾습니다.
    current = head
    while current is not None:
        if current.data == target:
            # Sol: 기존 다음 노드를 보존한 뒤 새 노드를 연결합니다.
            new_node = Node(name)
            new_node.next = current.next
            current.next = new_node
            return True
        current = current.next
    return False


def route_names(head):
    # Sol: None에 도달할 때까지 노드를 따라 이름을 모읍니다.
    result = []
    current = head
    while current is not None:
        result.append(current.data)
        current = current.next
    return result


names = input("초기 방문지(공백 구분): ").split()
target = input("기준 방문지: ")
name = input("새 방문지: ")
head = make_route(names)

# 두 함수의 반환값으로 결과를 출력합니다.
inserted = insert_after(head, target, name)
if not inserted:
    print("삽입할 위치가 없습니다.")
result = route_names(head)
if result:
    print(" -> ".join(result))
else:
    print("경로가 비어 있습니다.")
````

**문제 상황**

도서관 인쇄 관리 프로그램에 처리 되돌리기 기능을 추가합니다.
문서는 접수 순서대로 처리하지만, 되돌릴 때는 **가장 최근에 처리한 문서부터** 복구합니다.
이 문제의 되돌리기는 이미 나온 종이를 회수하는 작업이 아니라, 문서를 다시 처리할 대상으로 등록하는 동작입니다.

**처리 규칙**

- 대기열과 처리 이력은 처음에 모두 비어 있습니다.
- 접수한 문서는 대기열 맨 뒤에 넣습니다.
- 처리할 때는 대기열 맨 앞 문서를 꺼내 처리 이력에 기록합니다.
- 되돌릴 때는 처리 이력의 마지막 문서를 제거해 대기열 맨 앞에 넣습니다.
- 연속으로 되돌리면 최근 처리의 역순으로 복구됩니다. 복구한 문서를 다시 처리하면 새 처리 이력이 됩니다.
- 빈 대기열의 처리와 빈 이력의 되돌리기는 실패이며, 대기열과 이력을 모두 그대로 유지합니다.

**입출력 형식**

- **입력:** 첫 줄에 명령 수, 다음 줄부터 명령을 한 줄씩 입력합니다.
- `접수 문서명`: 접수만 하고 출력하지 않습니다.
- `처리`: 성공하면 `처리: 문서명`, 대기열이 비어 있으면 `대기 문서가 없습니다.`를 출력합니다.
- `되돌리기`: 성공하면 `복구: 문서명`, 처리 이력이 비어 있으면 `되돌릴 문서가 없습니다.`를 출력합니다.
- `확인`: 대기열 맨 앞을 제거하지 않고 `다음 문서: 문서명`을 출력합니다. 비어 있으면 `대기 문서가 없습니다.`를 출력합니다.
- **조건:** 명령 수는 1 이상 100 이하이며 명령 형식은 항상 올바릅니다. 문서명은 공백 없는 비어 있지 않은 문자열입니다. 같은 이름도 별개의 문서로 처리합니다.

**작성 안내**

**대기열 `queue`, 처리 이력 `history` 생성과 각 분기의 처리 및 출력**을 작성하세요.
`queue`는 `deque`, `history`는 리스트로 만들고, 이동하는 문서는 `document`에 저장합니다.
도전문제 1의 리스트 기반 스택 연산, 도전문제 2의 선입선출 규칙, 도전문제 3의 `deque` 연산을 조합합니다.
스택과 큐 클래스를 다시 만들 필요는 없습니다. 다른 문제 셀의 변수나 클래스에 의존하지 않습니다.

**출력 예시**

````python
# To Do
from collections import deque

# 빈 대기열 queue와 빈 처리 이력 history를 만드세요.


count = int(input("명령 수: "))
for i in range(count):
    parts = input("명령: ").split()
    command = parts[0]
    if command == "접수":
        # 새 문서를 대기열에 추가하세요.
        pass
    elif command == "처리":
        # 대기 문서를 처리하고 성공한 문서만 이력에 기록하세요.
        pass
    elif command == "되돌리기":
        # 가장 최근 처리 문서를 대기열로 복구하거나 빈 이력을 안내하세요.
        pass
    elif command == "확인":
        # 다음 처리 문서를 제거하지 않고 확인하세요.
        pass
````

````python
### Solution ###
from collections import deque

# Sol: 대기열은 큐로, 처리 이력은 스택으로 관리합니다.
queue = deque()
history = []

count = int(input("명령 수: "))
for i in range(count):
    parts = input("명령: ").split()
    command = parts[0]
    if command == "접수":
        # Sol: 새 문서를 대기열의 맨 뒤에 넣습니다.
        queue.append(parts[1])
    elif command == "처리":
        # Sol: 먼저 접수된 문서를 꺼내 처리 이력에 쌓습니다.
        if queue:
            document = queue.popleft()
            history.append(document)
            print(f"처리: {document}")
        else:
            print("대기 문서가 없습니다.")
    elif command == "되돌리기":
        # Sol: 가장 최근 처리 문서를 이력에서 꺼내 대기열 맨 앞에 복구합니다.
        if history:
            document = history.pop()
            queue.appendleft(document)
            print(f"복구: {document}")
        else:
            print("되돌릴 문서가 없습니다.")
    elif command == "확인":
        # Sol: 대기열을 바꾸지 않고 다음 문서를 보여 줍니다.
        if queue:
            print(f"다음 문서: {queue[0]}")
        else:
            print("대기 문서가 없습니다.")
````
