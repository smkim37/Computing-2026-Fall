````python
### Solution ###

def make_restock_report(items, stocks, limit):
    # 상품명과 재고를 묶고, 기준 이하인 상품만 딕셔너리에 담습니다.
    low_stock = {
        item: stock  # Sol: 상품명을 키, 재고를 값으로 저장합니다.
        for item, stock in zip(items, stocks)  # Sol: 같은 위치의 상품명과 재고를 묶습니다.
        if stock <= limit  # Sol: 재고가 기준 이하인지 확인합니다.
    }

    lines = []

    # 재입고할 상품에 1번부터 번호를 붙입니다.
    for number, (item, stock) in enumerate(low_stock.items(), start=1):  # Sol: 번호와 상품 정보를 함께 순회합니다.
        lines.append(f"{number}. {item}: {stock}개")  # Sol: 상품 하나의 출력 문장을 만듭니다.

    if len(lines) == 0:
        return "재입고할 상품이 없습니다."

    return "\n".join(lines)  # Sol: 문장들을 줄바꿈으로 연결합니다.


items = input("상품명을 입력하세요: ").split()
stocks = list(map(int, input("재고 수량을 입력하세요: ").split()))  # Sol: 재고를 정수 리스트로 변환합니다.
limit = int(input("재입고 기준을 입력하세요: "))  # Sol: 기준 수량을 정수로 변환합니다.

print(make_restock_report(items, stocks, limit))
````

````python
### Solution ###

def make_book_ranking(titles, votes, minimum_votes):
    # 도서명과 표 수를 같은 위치끼리 묶습니다.
    records = list(zip(titles, votes))  # Sol: 두 리스트의 같은 위치에 있는 값을 튜플로 묶습니다.

    # 최소 표 수 이상인 도서만 남깁니다.
    selected = list(filter(
        lambda record: record[1] >= minimum_votes,  # Sol: 표 수가 기준 이상인지 검사합니다.
        records,
    ))

    # 표 수 내림차순, 도서명 오름차순으로 정렬합니다.
    ranked = sorted(
        selected,
        key=lambda record: (-record[1], record[0]),  # Sol: 표 수에는 음수를 사용하고 도서명은 그대로 사용합니다.
    )

    lines = []

    # 정렬된 도서에 1위부터 순위를 붙입니다.
    for rank, (title, vote) in enumerate(ranked, start=1):  # Sol: 순위와 도서 정보를 함께 순회합니다.
        lines.append(f"{rank}위 {title} ({vote}표)")  # Sol: 도서 하나의 순위 문장을 만듭니다.

    if len(lines) == 0:  # Sol: 기준을 만족한 도서가 없는지 확인합니다.
        return "우선구매할 도서가 없습니다."

    return "\n".join(lines)  # Sol: 순위 문장들을 줄바꿈으로 연결합니다.


titles = input("도서명을 입력하세요: ").split()
votes = list(map(int, input("신청 표 수를 입력하세요: ").split()))  # Sol: 표 수를 정수 리스트로 변환합니다.
minimum_votes = int(input("최소 표 수를 입력하세요: "))  # Sol: 최소 표 수를 정수로 변환합니다.

print(make_book_ranking(titles, votes, minimum_votes))
````

### 도전문제 4. 안전한 조별 평균 계산기

<img src="../images/week2/week2_chal4.png" width="600" alt="조별 점수 평균을 계산하는 학생들">

**문제 상황**

한 조의 활동 점수들을 입력받아 평균을 계산하려고 합니다.  
정상 입력이면 평균을 소수점 첫째 자리까지 출력하고, 잘못된 입력은 예외 종류에 맞는 오류 문장으로 안내해 보세요.

**입출력 형식**
- **입력:** 공백으로 구분된 점수들을 한 줄에 입력
- **출력:** 정상 입력이면 평균, 잘못된 입력이면 오류 문장
- **조건:** 점수는 정수입니다. 정수가 아닌 값은 `ValueError`, 빈 입력은 평균 계산 중 발생하는 `ZeroDivisionError`로 처리합니다. `try-except-else`를 사용하세요.

**입력 예시**
```
80 90 100
```

**출력 예시**
```
점수를 입력하세요: 80 90 100
평균: 90.0점
```

````python
try:
    # map으로 입력값을 정수 리스트로 변환하세요.
    scores = ______

    # sum과 len을 사용해 평균을 계산하세요.
    average = ______ / ______
except ______:
    print("입력 오류: 점수는 정수로 입력하세요.")
except ______:
    print("입력 오류: 점수를 한 개 이상 입력하세요.")
else:
    print(f"평균: {______:.1f}점")
````

````python
### Solution ###

try:
    scores = list(map(int, input("점수를 입력하세요: ").split()))  # Sol: 입력값을 정수 리스트로 변환합니다.
    average = sum(scores) / len(scores)  # Sol: 점수 합계를 점수 개수로 나눕니다.
except ValueError:  # Sol: 정수가 아닌 입력에서 발생하는 예외를 처리합니다.
    print("입력 오류: 점수는 정수로 입력하세요.")
except ZeroDivisionError:  # Sol: 빈 점수 목록의 평균을 계산할 때 발생하는 예외를 처리합니다.
    print("입력 오류: 점수를 한 개 이상 입력하세요.")
else:
    print(f"평균: {average:.1f}점")  # Sol: 예외가 없을 때 평균을 소수점 첫째 자리까지 출력합니다.
````

````python
### Solution ###

try:
    teams = input("팀 이름을 입력하세요: ").split()
    numbers = list(map(int, input("배정 번호를 입력하세요: ").split()))  # Sol: 배정 번호를 정수 리스트로 변환합니다.
    if len(teams) == 0:  # Sol: 빈 팀 목록인지 확인합니다.
        raise ValueError("팀 이름을 한 개 이상 입력하세요.")
    if len(teams) != len(numbers):  # Sol: 팀과 번호의 개수가 같은지 확인합니다.
        raise ValueError("팀 수와 배정 번호 수가 같아야 합니다.")
    ordered = sorted(zip(teams, numbers), key=lambda pair: pair[1])  # Sol: 팀과 번호를 묶어 배정 번호순으로 정렬합니다.
    lines = [f"{order}. {team} ({number}번)" for order, (team, number) in enumerate(ordered, start=1)]  # Sol: 실제 발표 순서를 붙입니다.
except ValueError as error:
    print(f"입력 오류: {error}")
else:
    print("\n".join(lines))  # Sol: 발표 순서 문장을 줄바꿈으로 연결합니다.
````
