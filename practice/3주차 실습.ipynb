## 클래스 시작하기

화분마다 이름을 하나씩 저장한다고 생각해 봅시다.
**클래스**는 화분 정보를 어떻게 저장할지 정한 틀이고, **객체**는 그 틀로 만든 개별 화분 정보입니다.
특정 클래스로 만든 객체를 그 클래스의 **인스턴스**라고도 부릅니다.

**속성**은 객체가 가진 데이터이고, **메서드**는 클래스 안에 정의한 함수입니다.
아래 예제는 이름 속성 하나를 가진 객체를 만듭니다.

````python
class Plant:
    def __init__(self, name):
        self.name = name


plant = Plant("바질")
print(plant.name)
````

출력은 다음과 같습니다.

```
바질
```

| 코드 | 읽는 방법 |
| --- | --- |
| `class Plant:` | `Plant`라는 클래스를 정의합니다. 이 줄만으로 개별 화분 객체가 만들어지지는 않습니다. |
| `def __init__(self, name):` | 객체가 만들어질 때 이름을 받아 초기 상태를 설정합니다. |
| `self` | 지금 이 메서드가 다루는 객체 자신입니다. |
| `self.name = name` | 오른쪽의 전달받은 값을 왼쪽의 객체 속성에 저장합니다. |
| `plant = Plant("바질")` | 객체를 만들어 `plant`로 가리킵니다. `__init__`이 자동으로 실행되며, `self`를 직접 전달하지 않습니다. |
| `plant.name` | `plant`가 가리키는 객체에 저장된 이름을 읽습니다. |

`__init__`은 앞뒤에 밑줄이 **두 개씩** 있습니다.
클래스 안의 `def`는 네 칸, 그 메서드 안의 코드는 여덟 칸 들여씁니다.
객체를 만드는 마지막 두 줄은 클래스 밖이므로 들여쓰지 않습니다.

````python
### Solution ###

class CafeOrder:
    def __init__(self, menu, price, quantity):
        # Sol: 메뉴 이름, 가격, 수량을 이 주문의 속성에 저장합니다.
        self.menu = menu
        self.price = price
        self.quantity = quantity

    def total_price(self):
        # Sol: 저장된 가격과 수량으로 총 금액을 계산해 반환합니다.
        return self.price * self.quantity


# Sol: 메뉴 이름은 문자열로, 가격과 수량은 정수로 입력받습니다.
menu = input("메뉴 이름: ")
price = int(input("개당 가격(원): "))
quantity = int(input("주문 수량: "))

# Sol: 주문 객체를 만들고 메서드가 반환한 총 금액을 저장합니다.
order = CafeOrder(menu, price, quantity)
total = order.total_price()

# Sol: 객체에 저장된 메뉴와 수량, 계산한 총 금액을 출력합니다.
print(f"{order.menu} {order.quantity}개")
print(f"총 금액: {total}원")
````

### 도전문제 3. 잔여 매수를 지키는 인쇄 카드

**문제 상황**

도서관 인쇄 카드에는 사용할 수 있는 잔여 매수가 저장되어 있습니다.
인쇄할 때마다 잔여 매수가 줄어들지만, 잘못된 요청이나 잔여 매수를 초과한 요청은 거절해야 합니다.
같은 카드 객체에 여러 번 요청하고, 객체가 바뀐 잔여 매수를 계속 기억하도록 만들어 보세요.

**이번에 더할 개념: 상태 변경과 캡슐화**

문제 2의 메서드는 값을 계산해서 돌려주었습니다. 이번에는 메서드 안에서 속성 값 자체를 바꿉니다.
잔여 매수를 바꾸면 메서드 실행이 끝난 뒤에도 같은 카드 객체에 바뀐 값이 남습니다.

잔여 매수는 `self.__pages`에 저장합니다. 이름 앞의 밑줄 두 개는 클래스 내부에서 관리하려는 속성에 붙이는 표기이며,
외부에서 실수로 직접 접근하는 일을 줄여 줍니다.
이 문제에서는 바깥 코드가 잔여 매수를 직접 고치지 않고, **메서드를 통해 확인하고 변경**하도록 합니다.
이렇게 데이터와 처리 기능을 묶고 접근 방법을 정하는 것이 캡슐화의 기초입니다.

**만들 클래스**

| 항목 | 역할 |
| --- | --- |
| `PrintCard(name, pages)` | 이름을 `name`, 처음 잔여 매수를 `__pages`에 저장합니다. 이 부분은 제공됩니다. |
| `get_pages()` | 현재 잔여 매수를 반환합니다. 값을 바꾸지 않습니다. |
| `use(pages)` | 요청 매수가 1 이상이고 잔여 매수 이하일 때만 차감하고 `True`를 반환합니다. 나머지는 차감 없이 `False`를 반환합니다. |

`True`와 `False`는 성공 여부를 나타내는 불리언 값입니다. 반환값을 `if` 조건으로 사용하면 성공과 실패를 구분할 수 있습니다.
`get_pages()`와 `use()` 안에서는 출력하지 않습니다.

**입출력 형식**

- **입력:** 이름 한 줄, 처음 잔여 매수 한 줄, 요청 매수들을 공백으로 구분한 한 줄입니다. 총 3줄입니다.
- **출력:** 요청마다 `이름 | 요청장 사용 성공 | 잔여 남은매수장` 또는 `이름 | 요청장 사용 실패 | 잔여 남은매수장` 형식으로 한 줄 출력합니다.
- **조건:** 이름은 비어 있지 않습니다. 처음 잔여 매수는 0 이상의 정수입니다. 요청은 1개 이상이며 모두 정수입니다. 0 이하인 요청은 실패입니다. 남은 매수를 정확히 모두 사용하는 요청은 성공입니다.
- 실패한 요청은 잔여 매수를 바꾸지 않으며, 실패 후에도 다음 요청을 계속 처리합니다.

````python
### Solution ###

class PrintCard:
    # 제공 코드: 이름과 처음 잔여 매수를 저장합니다.
    def __init__(self, name, pages):
        self.name = name
        self.__pages = pages

    def get_pages(self):
        # Sol: 잔여 매수를 바꾸지 않고 현재 값을 반환합니다.
        return self.__pages

    def use(self, pages):
        # Sol: 사용할 수 없는 요청은 잔여 매수를 바꾸지 않고 거절합니다.
        if pages <= 0:
            return False
        if pages > self.__pages:
            return False
        # Sol: 사용할 수 있는 요청만 차감하고 성공을 반환합니다.
        self.__pages -= pages
        return True


# 제공 코드: 입력 부분은 그대로 사용하세요.
name = input("이름: ")
pages = int(input("처음 잔여 매수: "))
requests = list(map(int, input("요청 매수(공백 구분): ").split()))

# Sol: 반복문 앞에서 객체를 한 번만 만들어 잔여 매수를 계속 유지합니다.
card = PrintCard(name, pages)

for request in requests:
    # Sol: 같은 카드로 요청을 처리하고, 요청 후의 잔여 매수를 출력합니다.
    success = card.use(request)
    if success:
        result = "사용 성공"
    else:
        result = "사용 실패"
    print(f"{card.name} | {request}장 {result} | 잔여 {card.get_pages()}장")
````

````python
### Solution ###

# 제공 코드: 도전문제 3의 완성 코드입니다. 그대로 사용하세요.
class PrintCard:
    def __init__(self, name, pages):
        self.name = name
        self.__pages = pages

    def get_pages(self):
        return self.__pages

    def use(self, pages):
        if pages <= 0:
            return False
        if pages > self.__pages:
            return False
        self.__pages -= pages
        return True


class ReportCard(PrintCard):
    def print_report(self, pages_per_copy, copies):
        # Sol: 보고서 전체에 필요한 매수를 계산합니다.
        required_pages = pages_per_copy * copies
        # Sol: 상속받은 메서드로 차감하고 성공 여부를 그대로 반환합니다.
        return self.use(required_pages)


# 제공 코드: 입력 부분은 그대로 사용하세요.
name = input("이름: ")
pages = int(input("처음 잔여 매수: "))
pages_per_copy, copies = map(int, input("한 부 매수와 부수(공백 구분): ").split())

# Sol: 부모의 초기화 메서드를 사용하는 자식 객체로 보고서 인쇄를 요청합니다.
card = ReportCard(name, pages)
success = card.print_report(pages_per_copy, copies)

# Sol: 인쇄 성공 여부와 요청 후 잔여 매수를 출력합니다.
if success:
    print(f"{card.name}: 인쇄 성공")
else:
    print(f"{card.name}: 인쇄 실패")
print(f"잔여: {card.get_pages()}장")
````

````python
### Solution ###

class SnackBox:
    def __init__(self, snack, stock):
        # Sol: 간식 이름과 이 상자의 내부 수량을 저장합니다.
        self.snack = snack
        self.__stock = stock

    def get_stock(self):
        # Sol: 간식 개수를 바꾸지 않고 현재 값을 반환합니다.
        return self.__stock

    def take(self, count):
        # Sol: 잘못된 요청이나 간식이 부족한 요청은 차감 없이 거절합니다.
        if count <= 0:
            return False
        if count > self.__stock:
            return False
        # Sol: 나눌 수 있는 개수만 차감하고 성공을 반환합니다.
        self.__stock -= count
        return True


class SnackSetBox(SnackBox):
    def share_sets(self, items_per_set, sets):
        # Sol: 꾸러미 전체에 필요한 간식 수를 계산합니다.
        required_items = items_per_set * sets
        # Sol: 상속받은 메서드로 차감하고 성공 여부를 그대로 반환합니다.
        return self.take(required_items)


# 제공 코드: 입력 부분은 그대로 사용하세요.
snack = input("간식 이름: ")
stock = int(input("처음 간식 개수: "))
items_per_set = int(input("꾸러미당 간식 수: "))
requests = list(map(int, input("요청별 꾸러미 수(공백 구분): ").split()))

# Sol: 반복문 앞에서 객체를 한 번만 만들어 남은 간식 수를 계속 유지합니다.
box = SnackSetBox(snack, stock)

for sets in requests:
    # Sol: 같은 상자로 꾸러미 나눔을 요청하고, 요청 후 남은 개수를 출력합니다.
    success = box.share_sets(items_per_set, sets)
    if success:
        result = "나눔 성공"
    else:
        result = "나눔 실패"
    print(f"{sets}꾸러미 {result} | 남은 {box.snack}: {box.get_stock()}개")
````
