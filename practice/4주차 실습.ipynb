### 도전문제 1. 물품 코드의 확인 숫자


<img src="../images/week4/week4_chal1.png" width="600" alt="대여 물품 상자에 붙은 식별 코드를 확인하는 모습">


**문제 상황**

교내 장비실에서는 물품 코드가 잘못 입력되었는지 확인하는 데 사용할 숫자를 만듭니다.
코드의 각 자릿수를 모두 더한 뒤, 그 합의 **일의 자리 숫자**를 확인 숫자로 사용합니다.
예를 들어 코드가 `5724`이면 자릿수의 합은 18이고 확인 숫자는 8입니다.
합이 두 자리 이상이어도 자릿수를 다시 더하지 않습니다.

자릿수의 합을 구하는 부분을 재귀함수로 작성해 보세요.

**만들 함수**

| 함수 | 역할 |
| --- | --- |
| `digit_sum(n)` | 0 이상의 정수 `n`의 자릿수 합을 정수로 반환합니다. `n`이 0이면 0을 반환합니다. |

**입출력 형식**

- **입력:** 물품 코드 한 줄입니다.
- **출력:** `확인 숫자: 숫자` 형식으로 한 줄 출력합니다.
- **조건:** 코드는 0 이상 999999999999 이하의 정수입니다. 0을 제외한 코드의 맨 앞에는 0이 없습니다.
- `digit_sum()`은 반드시 자기 자신을 호출하는 재귀함수로 작성합니다. 반복문, 문자열 변환, `sum()`은 사용하지 않습니다.

**작성 안내**

아래 `To Do` 셀에서 **함수 정의, 입력, 함수 호출, 출력까지 전체 코드**를 작성하세요.
입력은 `number`에 저장하고, 입력 안내 문구는 `물품 코드: `를 사용합니다.
함수가 반환한 자릿수 합으로 확인 숫자를 구해 `check`에 저장한 뒤 출력합니다.

재귀 호출 없이 답을 반환할 수 있는 경우를 먼저 정하고, 호출할 때마다 처리할 자릿수가 줄어들도록 만드세요.
`digit_sum()` 안에서는 합을 반환하고, 입력과 출력은 함수 밖에서 처리합니다.

**입력 예시**
```
5724
```

**출력 예시**
```
물품 코드: 5724
확인 숫자: 8
```


````python
# To Do

# 자릿수의 합을 반환하는 재귀함수 digit_sum(n)을 정의하세요.


# 물품 코드를 입력받아 number에 저장하세요.


# 재귀함수의 반환값으로 확인 숫자 check를 구하고 출력하세요.
````

````python
### Solution ###

def digit_sum(n):
    # Sol: 한 자리 수는 그 숫자 자체가 자릿수의 합입니다.
    if n < 10:
        return n
    # Sol: 마지막 자릿수와 나머지 자릿수의 합을 더해 반환합니다.
    return n % 10 + digit_sum(n // 10)


# Sol: 물품 코드를 정수로 입력받습니다.
number = int(input("물품 코드: "))

# Sol: 자릿수 합의 일의 자리를 확인 숫자로 출력합니다.
check = digit_sum(number) % 10
print(f"확인 숫자: {check}")
````

문자열 인덱스로 글자를 읽기 전에, 더 비교할 글자가 있는지 확인하세요.

문자열에도 인덱스와 슬라이싱을 사용할 수 있습니다.
`text[0]`은 첫 글자, `text[-1]`은 마지막 글자, `text[1:-1]`은 양 끝 글자를 제외한 부분입니다.
한 번의 호출에서 처리할 부분과 다음 호출에 맡길 부분을 나누어 보세요.

````python
# To Do

def is_palindrome(text):
    # 종료 조건과 재귀 호출을 사용해 대칭 여부를 반환하세요.
    pass


# 제공 코드: 입력 부분은 그대로 사용하세요.
code = input("입장 코드: ")

# 재귀함수의 반환값에 따라 대칭 여부를 출력하세요.
````

````python
### Solution ###

def is_palindrome(text):
    # Sol: 글자가 없거나 하나이면 더 비교할 쌍이 없습니다.
    if len(text) <= 1:
        return True
    # Sol: 양 끝이 다르면 나머지를 검사하지 않고 종료합니다.
    if text[0] != text[-1]:
        return False
    # Sol: 양 끝을 제외한 부분의 검사 결과를 그대로 반환합니다.
    return is_palindrome(text[1:-1])


# 제공 코드: 입력 부분은 그대로 사용하세요.
code = input("입장 코드: ")

# Sol: 재귀함수의 불리언 반환값에 따라 결과를 출력합니다.
if is_palindrome(code):
    print("대칭 코드입니다.")
else:
    print("대칭 코드가 아닙니다.")
````

````python
### Solution ###

def split_ribbon(length, limit):
    # Sol: 더 자르지 않는 리본은 최종 조각 한 개로 처리합니다.
    if length <= limit:
        print(f"조각: {length}cm")
        return 1

    # Sol: 홀수일 때 앞쪽이 1cm 짧도록 두 조각의 길이를 구합니다.
    front = length // 2
    back = length - front

    # Sol: 앞쪽을 모두 처리한 뒤 뒤쪽을 처리하고 각 조각 수를 저장합니다.
    front_count = split_ribbon(front, limit)
    back_count = split_ribbon(back, limit)

    # Sol: 두 호출이 반환한 조각 수를 합쳐 반환합니다.
    return front_count + back_count


# 제공 코드: 입력 부분은 그대로 사용하세요.
length = int(input("리본 길이(cm): "))
limit = int(input("최대 조각 길이(cm): "))

# Sol: 함수가 출력한 최종 조각들의 총 개수를 마지막에 출력합니다.
count = split_ribbon(length, limit)
print(f"총 조각: {count}개")
````

````python
### Solution ###

def digit_sum(n):
    # Sol: 한 자리 수는 그 숫자 자체를 반환합니다.
    if n < 10:
        return n
    # Sol: 마지막 자릿수에 나머지 자릿수의 합을 더합니다.
    return n % 10 + digit_sum(n // 10)


def is_palindrome(text):
    # Sol: 비교할 쌍이 남지 않으면 대칭입니다.
    if len(text) <= 1:
        return True
    # Sol: 양 끝이 다르면 즉시 대칭이 아니라고 반환합니다.
    if text[0] != text[-1]:
        return False
    # Sol: 양 끝을 제외한 부분의 검사 결과를 반환합니다.
    return is_palindrome(text[1:-1])


# 제공 코드: 입력 부분은 그대로 사용하세요.
code = input("입장 코드: ")
expected = int(input("입장권 확인 숫자: "))

# Sol: 대칭 검사는 원래 문자열로, 자릿수 합 계산은 정수로 처리합니다.
symmetric = is_palindrome(code)
check = digit_sum(int(code)) % 10

# Sol: 승인 여부와 관계없이 대칭 여부와 계산한 확인 숫자를 출력합니다.
if symmetric:
    print("대칭 여부: 예")
else:
    print("대칭 여부: 아니요")
print(f"계산한 확인 숫자: {check}")

# Sol: 두 조건을 모두 만족할 때만 입장을 승인합니다.
if symmetric and check == expected:
    print("입장: 승인")
else:
    print("입장: 거절")
````
