````python
### Solution ###

class Book:
    def __init__(self, key, title, available):
        self.key = key
        self.title = title
        self.available = available


def binary_search(A, key, low, high):
    if low <= high:    # Sol: 탐색할 범위가 남아 있을 때만 확인
        middle = (low + high) // 2    # Sol: 현재 범위의 가운데 인덱스 계산
        if key == A[middle].key:    # Sol: 가운데 도서 번호와 찾을 번호가 같은지 확인
            return middle    # Sol: 찾은 도서의 인덱스를 반환
        elif key < A[middle].key:    # Sol: 찾을 번호가 더 작으면 왼쪽 범위 선택
            return binary_search(A, key, low, middle - 1)    # Sol: 왼쪽 탐색 결과를 반환
        else:
            return binary_search(A, key, middle + 1, high)    # Sol: 오른쪽 탐색 결과를 반환
    return None    # Sol: low가 high보다 커지면 탐색할 항목이 없으므로 실패


n = int(input("도서 수: "))
A = []
for _ in range(n):
    number, title, status = input("도서 번호, 제목, 대출 상태: ").split()
    A.append(Book(int(number), title, status == "1"))

key = int(input("찾을 도서 번호: "))

index = binary_search(A, key, 0, len(A) - 1)    # Sol: 전체 도서 목록에서 탐색
if index is not None:    # Sol: 인덱스 0을 포함한 탐색 성공 처리
    print(f"도서: {A[index].title}")    # Sol: 찾은 도서의 제목 출력
    if A[index].available:    # Sol: 등록된 도서의 대출 가능 여부 확인
        print("대출 가능")    # Sol: 대출 가능한 도서 안내
    else:
        print("대출 중")    # Sol: 이미 대출 중인 도서 안내
else:
    print("등록되지 않은 도서입니다.")    # Sol: 번호가 없는 경우 안내
````

````python
### Solution ###

def hash_insert(table, key):
    M = len(table)    # Sol: 테이블의 칸 수를 해시 주소 계산에 사용
    i = key % M    # Sol: 물품 번호의 해시 주소 계산
    while table[i] is not None:    # Sol: 이미 물품이 있으면 다음 칸을 조사
        i = (i + 1) % M    # Sol: 마지막 칸 다음에는 0번 칸으로 이동
    table[i] = key    # Sol: 처음 만난 빈칸에 물품 번호 저장


def hash_search(table, key):
    M = len(table)    # Sol: 삽입할 때와 같은 테이블 크기 사용
    i = key % M    # Sol: 삽입할 때의 시작 주소에서 탐색
    while table[i] is not None:    # Sol: 빈칸을 만나기 전까지 확인
        if table[i] == key:    # Sol: 현재 칸의 물품 번호와 비교
            return i    # Sol: 찾은 칸의 인덱스 반환
        i = (i + 1) % M    # Sol: 충돌로 뒤에 저장된 물품까지 순환 조사
    return None    # Sol: 빈칸에 도달하면 등록되지 않은 번호


M = 13
table = [None] * M
numbers = list(map(int, input("물품 번호(공백 구분): ").split()))
target = int(input("찾을 물품 번호: "))

for key in numbers:    # Sol: 입력된 순서대로 물품 번호를 저장
    hash_insert(table, key)    # Sol: 충돌을 처리하며 테이블에 삽입

index = hash_search(table, target)    # Sol: 요청한 번호의 저장 위치 탐색
if index is not None:    # Sol: 0번 칸을 포함한 탐색 성공 처리
    print(f"보관 칸: {index}")    # Sol: 0부터 시작하는 실제 칸 번호 출력
else:
    print("등록되지 않은 물품입니다.")    # Sol: 탐색에 실패했을 때 안내
````

````python
### Solution ###

class Book:
    def __init__(self, key, title, available):
        self.key = key
        self.title = title
        self.available = available


def binary_search(A, key, low, high):
    if low <= high:
        middle = (low + high) // 2
        if key == A[middle].key:
            return middle
        elif key < A[middle].key:
            return binary_search(A, key, low, middle - 1)
        else:
            return binary_search(A, key, middle + 1, high)
    return None


n = int(input("도서 수: "))
A = []
for _ in range(n):
    number, title, status = input("도서 번호, 제목, 대출 상태: ").split()
    A.append(Book(int(number), title, status == "1"))

q = int(input("조회 요청 수: "))
requests = []
for _ in range(q):
    requests.append(int(input("조회할 도서 번호: ")))

count = 0    # Sol: 대출 가능한 조회 요청 수를 누적할 변수
for key in requests:    # Sol: 입력된 순서대로 요청을 하나씩 처리
    index = binary_search(A, key, 0, len(A) - 1)    # Sol: 매 요청마다 전체 목록에서 탐색
    if index is None:    # Sol: 번호가 없으면 도서 속성에 접근하지 않음
        print(f"{key}: 미등록")    # Sol: 등록되지 않은 번호 안내
    elif A[index].available:    # Sol: 찾은 도서가 대출 가능한지 확인
        print(f"{key}: 대출 가능")    # Sol: 이번 요청의 조회 결과 출력
        count += 1    # Sol: 같은 번호의 반복 요청도 각각 1회로 계산
    else:
        print(f"{key}: 대출 중")    # Sol: 등록되어 있지만 대출 중인 도서 안내
print(f"대출 가능한 요청 수: {count}")    # Sol: 대출 가능한 요청의 합계 출력
````
