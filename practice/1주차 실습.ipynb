````python
### Solution ###

def standardize_phone_number(phone):
    digits = ""

    # phone을 처음부터 끝까지 살펴보며 숫자만 digits에 더합니다.
    for index in range(len(phone)):
        character = phone[index]
        if character.isdigit():  # Sol: 현재 문자가 숫자인지 확인합니다.
            digits += character  # Sol: 숫자만 이어 붙입니다.

    # 숫자가 11개가 아니거나 010으로 시작하지 않으면 잘못된 번호입니다.
    if len(digits) != 11 or digits[:3] != "010":  # Sol: 길이와 시작 번호를 검사합니다.
        return None

    # 숫자 사이에 하이픈을 넣어 일정한 형식으로 반환합니다.
    return f"{digits[:3]}-{digits[3:7]}-{digits[7:]}"  # Sol: 세 구간을 하이픈으로 연결합니다.


phone = input("전화번호를 입력하세요: ")
result = standardize_phone_number(phone)

if result is None:  # Sol: 잘못된 번호일 경우 None을 반환하므로 이를 확인합니다.
    print("잘못된 전화번호입니다. 010으로 시작하는 숫자 11개를 입력하세요.")
else:
    print(result)
````

````python
### Solution ###

MENU_PRICES = {
    "아메리카노": 3000,
    "카페라테": 4000,
    "레몬에이드": 4500,
}


def calculate_total(menu, quantity, has_student_discount):
    # 딕셔너리에서 메뉴 가격을 찾고 수량을 곱합니다.
    subtotal = MENU_PRICES[menu] * quantity  # Sol: 메뉴 가격과 수량을 곱합니다.

    # 학생 할인이 적용되면 주문 금액의 90%를 정수로 계산합니다.
    if has_student_discount:  # Sol: 학생 할인 적용 여부를 확인합니다.
        subtotal = subtotal * 90 // 100  # Sol: 10% 할인된 금액을 계산합니다.

    return subtotal


menu = input("메뉴를 입력하세요: ")
quantity = int(input("수량을 입력하세요: "))
has_student_discount = input("학생 할인을 적용할까요? (Y/N): ") == "Y"

total = calculate_total(menu, quantity, has_student_discount)
discount_text = "적용" if has_student_discount else "미적용"

print(f"메뉴: {menu}")
print(f"수량: {quantity}개")
print(f"학생 할인: {discount_text}")
print(f"최종 금액: {total}원")
````

````python
### Solution ###

def has_only_even_digits(number):
    # 도전문제 1에서 사용한 //와 %로 각 자릿수를 구합니다.
    hundreds = number // 100  # Sol: 백의 자리를 구합니다.
    tens = (number // 10) % 10  # Sol: 십의 자리를 구합니다.
    ones = number % 10  # Sol: 일의 자리를 구합니다.

    # 홀수인 자리가 하나라도 있으면 조건을 만족하지 않습니다.
    if hundreds % 2 != 0:  # Sol: 백의 자리가 홀수인지 확인합니다.
        return False
    if tens % 2 != 0:  # Sol: 십의 자리가 홀수인지 확인합니다.
        return False
    if ones % 2 != 0:  # Sol: 일의 자리가 홀수인지 확인합니다.
        return False

    return True


start_number = int(input("시작 번호를 입력하세요: "))
end_number = int(input("끝 번호를 입력하세요: "))

matching_numbers = ""

# 도전문제 2와 같이 range를 사용해 범위를 처음부터 끝까지 확인합니다.
for number in range(start_number, end_number + 1):
    if has_only_even_digits(number):
        if matching_numbers != "":
            matching_numbers += ","
        matching_numbers += f"{number}"

print(matching_numbers)
````
